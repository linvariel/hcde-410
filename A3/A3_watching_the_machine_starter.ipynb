The probe I used involves searching for and synthesizing King, Yakima, and Whitman County's 2026 August primary election voter turnout results. The LLM model I used for the first three probes is Claude Sonnet 5.5 Medium and ChatGPT GPT-6 for the fourth probe, using the following prompt: 

> Find the voter turnout in King, Whitman, and Yakima counties in the August 2026 primary election, calculated as ballots cast divided by registered voters. Give a percent rounded to the nearest whole number and provide data sources.

This prompt directly asks for the turnout (as attaching a dataset would require downloading and attaching additional files that those who I observed in Part 3 preferred not to do), which tests the LLM's ability to search for multiple correct sources and data. The specific detail in the prompt about rounding to whole numbers tests the LLM's ability to adhere to formatting requests.

I verified the results of these LLMs by ensuring the reported whole number percentages matched official voter turnout results displayed on the respective county election result webpages:
- [King County Election Results](https://app.enhancedvoting.com/results/public/king-county-wa/elections/20260804)
- [Whitman County Election Results](https://app.enhancedvoting.com/results/public/whitman-county-wa/elections/20260804)
- [Yakima County Election Results](https://app.enhancedvoting.com/results/public/yakima-county-wa/elections/20260804)

I also explored cited sources to find relevant information that the responses claimed to have used as sources. For example, I found the section in the cited webpage where ChatGPT got its election-day turnout percentages.

**Same prompt, same LLM**

Variations in these responses included length of responses, the formatting of sources (e.g., whether incorporated into explanations as links, or included separately in bullet point format for each source). County, ballots cast, registered voters, and turnout were consistently formatted in a table, with variations in how the turnout column was reported: sometimes as percentages with single or double decimal places followed by an arrow and a bolded whole number percentage, percentages with single double decimal places followed by "so" and a bolded whole number percentage, or unbolded whole number percentages with the percentages with decimal places in parentheses.

**Same prompt, different LLM**

For the one ChatGPT response, the response was fairly different from the Claude responses. The first sentence reported the overall voter turnout in Washington state. These reported turnouts did not match the final turnouts from the respective county turnout webpages I used to verify responses, but I did find the section of a website ChatGPT linked under the table where it likely pulled its non-final, election-day turnout percentages.
